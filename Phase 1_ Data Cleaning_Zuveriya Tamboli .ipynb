```python
orders = pd.read_csv("Data/orders DA.csv")
order_items = pd.read_csv("Data/order_items DA.csv")
customers = pd.read_csv("Data/customers DA.csv")
inventory = pd.read_csv("Data/inventory DA.csv")
payments = pd.read_csv("Data/payments DA.csv")
products = pd.read_csv("Data/products DA.csv")
suppliers = pd.read_csv("Data/suppliers DA.csv")
returns = pd.read_csv("Data/returns DA.csv")

print("all 8 csv files loaded suessfully")
```

#Step 1 : Check the no of rows and columns

```python
datasets = {
    "Orders": orders,
    "Order Items": order_items,
    "Customers": customers,
    "Products": products,
    "Payments": payments,
    "Returns": returns,
    "Inventory": inventory,
    "Suppliers": suppliers
}
for name ,df in datasets.items():
    
    print(name, ":", df.shape[0], "rows and", df.shape[1], "columns")
```

```python
# Step 2 : Check col names and data types
```

```python
for name,df in datasets.items():
    print(f"\n{'='*50}")
    print(f"dataset:{name}")
    print("="*50)
    
    print("\nColumns names:")
    print(df.columns.tolist())
    
    print("\n datatypes:")
    print(df.dtypes)
```

```python
# Step 3 check missing values in all 8 datasets
for name, df in datasets.items():
    print(f"/n--- {name}---")
    missing = df.isnull().sum()
    print(missing[missing >0] if missing.sum() >0
         else " no missing value found")
    print("total missing values", missing.sum())
```

```python
# check order status against missing delivery dates
```

```python
print("orders: delivery status vs missing delivery dates")
```

```python
print(orders[orders["delivered_date"].isnull()]["status"].value_counts())
print(payments.isnull().sum())
```

```python
for name, df in datasets.items():
    print(name,":",df.duplicated().sum(),"duplicate rows")
```

```python
# step 6 : check for duplicate id
```

```python

id_columns = {
    "Orders": ["order_id"],
    "Order Items": ["order_item_id"],
    "Customers": ["customer_id"],
    "Products": ["product_id"],
    "Payments": ["payment_id"],
    "Returns": ["return_id"],
    "Inventory": ["inventory_id"],
    "Suppliers": ["supplier_id"]
}

for name, cols in id_columns.items():
    df = datasets[name]
    for col in cols:
        if col in df.columns:
            print(name, col, ":", df[col].duplicated().sum(), "duplicate IDs")
```

```python
#step 7 : check for date columns 
```

```python
for name,df in datasets.items():
    print(f"\n{name}")
    for col in df.columns:
        if "date" in col.lower():
            print(col,":",df[col].dtype)
```

```python
# step 8 convert date colum
```

```python

date_columns = {
    "Orders": ["order_date", "delivered_date"],
    "Customers": ["registration_date", "last_login_date"],
    "Products": ["launch_date"],
    "Payments": ["payment_date", "refund_date"],
    "Returns": ["return_date", "refund_date"],
    "Inventory": ["last_restocked_date"],
    "Suppliers": ["created_date"]
}

for name, cols in date_columns.items():
    for col in cols:
        datasets[name][col] = pd.to_datetime(
            datasets[name][col], errors="coerce"
        )

print("Date conversion completed!")
```

```python

for name, df in datasets.items():
    for col in df.columns:
        if "date" in col.lower():
            print(name, col, ":", df[col].dtype)
```

```python
# step 9 check numeric col
```

```python

for name, df in datasets.items():
    print(f"\n{name}")
    print(df.select_dtypes(include="object").columns.tolist())
```

```python

for name, df in datasets.items():
    print(f"\n{name}")
    print(df.select_dtypes(include="number").columns.tolist())
```

```python

check_columns = {
    "Orders": ["pincode", "is_cod"],
    "Customers": ["phone", "pincode", "is_verified", "is_active"],
    "Products": ["hsn_code", "is_active"],
    "Suppliers": ["phone", "pincode", "is_active"]
}

for name, cols in check_columns.items():
    for col in cols:
        if col in datasets[name].columns:
            print(name, "|", col, "|", datasets[name][col].dtype)
```

```python
#Step 12: Correct identifier and flag columns
```

```python

text_columns = {
    "Orders": ["pincode"],
    "Customers": ["phone", "pincode"],
    "Products": ["hsn_code"],
    "Suppliers": ["phone", "pincode"]
}

for name, cols in text_columns.items():
    for col in cols:
        datasets[name][col] = datasets[name][col].astype("string")

print("Identifier columns converted to text!")
```

```python
for name, cols in text_columns.items():
    for col in cols:
        print(name, col, datasets[name][col].dtype)
```

```python

for name, df in datasets.items():
    print(
        name,
        "| Rows:", len(df),
        "| Missing values:", df.isnull().sum().sum(),
        "| Duplicate rows:", df.duplicated().sum()
    )
```
